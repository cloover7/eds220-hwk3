````python
# FINAL CODE

# import libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# read in data
fp = os.path.join('data','tl_2026_us_county','tl_2026_us_county.shp')
counties = gpd.read_file(fp)

fp1 = 'data/Spill_Incidents_20261104.csv'
spills = pd.read_csv(fp1)

# rename lengthy column names
spills = spills.rename(columns={
    'Spill Number': 'Spill_ID',
    'Program Facility Name': 'Facility',
    'Material Name': 'Material'
})

# convert date columns to datetime format
spills['Spill Date'] = pd.to_datetime(spills['Spill Date'])
spills['Close Date'] = pd.to_datetime(spills['Close Date'])
spills['Received Date'] = pd.to_datetime(spills['Received Date'])

# select 2023 petroleum spills
petr_23 = spills[(spills['Spill Date'].between('2023-1-1', '2023-12-31')) & (spills['Material Family'] == 'Petroleum')]

# ensure data is petroleum spills only
assert petr_23['Material Family'].unique() == 'Petroleum'

# aggregate spill data by county
spills_per_county = petr_23.groupby('County')['Spill_ID'].nunique().reset_index()
spills_per_county = spills_per_county.rename(columns = {
    'County': 'county',
    'Spill_ID':'n_spills'
})


# select only new york counties
counties[counties['NAME'] == 'New York']
ny_counties = counties[counties['STATEFP'] == '36']

# harmonize St. Lawrence
spills_per_county.iloc[52, 0] = 'St. Lawrence'
assert spills_per_county['county'].isin(['St. Lawrence']).any()

# merge counties and spills data
ny_spills = pd.merge(ny_counties, spills_per_county, left_on='NAME', right_on='county', how='inner')

# update niagara and onondaga with reserve spills
ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'] += 2
ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'] += 1
ny_spills

# assert update worked properly
assert ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'].item() == 99
assert ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'].item() == 215


# map spill data
fig, ax = plt.subplots(figsize=(9, 5)) # Update figure size

# Remove the axis for a cleaner map
ax.axis('off')

# Title for the plot
ax.set_title('Petroleum Spills Per New York County in 2023', 
              fontsize=15)

# add annotation
fig.text(0.1, 0.1, 'Data Source: New York State Department of Environmental Conservation \n Access Date: Nov 5, 2026', fontsize=7)

# Add states
ny_spills.plot(ax=ax, column = 'n_spills', cmap='Reds', legend = True, legend_kwds = {'label':'Number of Spills'})

plt.show()

````
